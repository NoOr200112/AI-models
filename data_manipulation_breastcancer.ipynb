```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv("breast-cancer-wisconsin.csv")

# --- Display basic info ---
print("First 5 rows:")
print(data.head(), "\n")

print("Dataset Info:")
print(data.info(), "\n")

print("Summary Statistics:")
print(data.describe(), "\n")

# --- Check for missing values ---
print("Missing Values:")
print(data.isnull().sum(), "\n")
```

```python
import numpy as np

# Replace '?' with NaN
data = data.replace('?', np.nan)

# Verify the replacement
print("Missing Values after replacement:")
print(data.isna().sum())
```

```python
# Convert column '1.3' to numeric, coercing errors to NaN
data['1.3'] = pd.to_numeric(data['1.3'], errors='coerce')

# Replace missing values with the mean of the column using fillna()
data.fillna(data.mean(numeric_only=True), inplace=True)

# Verify that there are no more missing values
print("Missing Values after filling with mean:")
print(data.isnull().sum())
```

```python
# Display the values of the column that had missing values after filling
print("\nValeurs de la colonne '1.3' après remplacement des valeurs manquantes:")
print(data['1.3'].head())

# You can refer to the output of the previous cell to see the count of missing values before replacement.
```

```python
# Store the number of rows before dropping
rows_before_drop = data.shape[0]
print(f"Nombre d'échantillons avant dropna(): {rows_before_drop}")

# Drop rows with missing values using dropna()
data_dropped = data.dropna()

# Store the number of rows after dropping
rows_after_drop = data_dropped.shape[0]
print(f"Nombre d'échantillons après dropna(): {rows_after_drop}")
```
