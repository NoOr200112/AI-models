```python
import pandas as pd

# Load the CSV file
file_path = 'Solardatanew.csv'
data = pd.read_csv(file_path)

# Display the first few rows of the dataframe to inspect it
data.head()
```

```python
# Split the single column into 8 separate columns based on commas
data_split = data.iloc[:, 0].str.split(',', expand=True)

# Assign proper column names based on the original structure
data_split.columns = [
    "Timestamp", "temperature", "humidity", "irradiation",
    "RTD", "pvVoltage", "pvCurrent", "Power_Generated"
]

# Display the first few rows of the newly split dataframe
data_split.head()
```

```python
# Save the updated dataframe to a new CSV file
output_file_path = 'Solardatanew_split.csv'
data_split.to_csv(output_file_path, index=False)

output_file_path
```
